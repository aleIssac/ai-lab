# 03 发布成页面

**目标**：把 RAG 链路包成业务人员能直接用的页面。

`apps/rag_app.py` 已经写好，这里只演示怎么起、怎么验。

## 1. 启动

在**终端**（不是 notebook）执行：

```bash
streamlit run apps/rag_app.py --server.address 0.0.0.0 --server.port 8502
```

然后打开 http://localhost:8502 。

> 注意 `--server.address 0.0.0.0`：默认只监听 localhost，容器外访问不到。

## 2. 页面能做什么

| 区域 | 功能 |
|---|---|
| 侧边栏 | 填 API Key、选模型、调检索片段数 |
| 主区 | 提问 → 答案 + 引用来源 |
| 右侧 | 展开每个检索片段看原文和相似度 |

**没填 Key 也能演示**：页面会展示「检索到了什么」和「拼好的 prompt」，
这反而更适合讲清楚 RAG 的结构。

## 3. 健康检查

````python
import requests

try:
    r = requests.get('http://localhost:8502/_stcore/health', timeout=5)
    print('Streamlit 状态:', r.status_code, r.text[:50])
except Exception as e:
    print('服务未启动:', type(e).__name__)
    print('先在终端执行 streamlit run apps/rag_app.py --server.port 8502')
````

## 小结

- 部署的本质是让不懂模型的人也能用模型，所以交互设计比模型参数重要
- 这一条链路（文档 → 切块 → 向量化 → 检索 → 生成 → 页面）就是大模型应用的标准形态
- 到这里，需求④的三个专项（AI 应用 / 大数据 / 大模型）全部闭环
